```python
def sq(banana):
    return banana * banana
```

```python
numbers = list(range(1,11))
numbers
```

```python
# list(map(sq, numbers))
list(map(lambda banana : banana * banana , numbers))
```

```python
week = ["mon", "tue","wed"]
# use lambda and map and capitalize these strings
days = list(map(lambda x : x.capitalize(), week))
days
```

```python
list(map(lambda x : "Even" if x % 2 == 0 else "Odd" , numbers))
```

```python
# if number is greater than 5 multiply it by 100 else leave the number
# as it is
list(map(lambda x : x*100 if x > 5 else x, numbers))
```

```python
scores = [55, 78, 92, 40, 60, 85]
# if the score is 60 or above ---> pass fail
```

```python
list(map(lambda x : "Pass" if x >= 60 else "fail" , scores))
```

```python
names = ["Anna", "Alexander", "Bo", "Catherine"]
```

```python
# if name is more than 5 letter ---> Long else short
list(map(lambda x: 'Long' if len(x) > 5  else 'Short', names))
```
