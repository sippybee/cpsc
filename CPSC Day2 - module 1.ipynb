````python
def isvowel(letter):
    """
    Returns True if letter is a vowel
    Letter ---> str
    """
    letter = letter.lower()
    vowels = ("a","u","i","e","o")
    if letter in vowels:
        return True
    else:
        return False

def game(word):
    if isvowel(word[0]):
        new_word = word + "ay"
    else:
        for i in range(len(word)):
            if isvowel(word[i]):
                print(i, word[i])
                part_one = word[0:i]
                part_two = word[i:]
                new_word = part_two + part_one + "yay"
                break
    return new_word

def main():
    word = input("Give me a word ")
    if word.isalpha() and len(word)>1:
        result = game(word)
    else:
        result = "Not a valid input"
    return result
````

```
Pig Latin challenge:
Pig Latin (also known as Dog Latin) is a language game in which words are altered according to certain rules. To make the pig latin form of an English word, the initial consonant sound is transposed to the end of the word, and an “ay” is affixed. Specifically, there are two rules:
If a word begins with a vowel, append “yay” to the end of the word. If a word begins with a consonant, remove all the consonants from the beginning up to the first vowel and append them to the end of the word. Finally, append “ay” to the end of the word.
For example:
dog => ogday
python => onpythay
scratch => atchscray
is => isyay
apple => appleyay
```

````python
s = "what time do i have to be there? what's the address? see you then!"
s = s.capitalize()
s = s.replace(" i ", " I ")
pun = ["!","?","."]

for i in range(len(s)-2):
    if s[i] in pun:
        inx_lttr_chang = i + 2
        letter_change = s[inx_lttr_chang]
        new_letter = letter_change.upper()
        part_one = s[0:inx_lttr_chang]
        part_two = s[inx_lttr_chang+1 : ]
#         print(part_two)
        s = f"{part_one}{new_letter}{part_two}"
s
````

````python
def cleaning(w):
    new = ""
    for i in w:
        if i.isalnum():
            new +=i
    return new

def get_number(t):
    return t[1]

obj = open(r"article.txt", encoding="utf8")
data = obj.read().lower().split()
d = {}
for word in data:
    word = cleaning(word)
    if word in d:
        d[word]+=1
    else:
        d[word] = 1
alist = list(d.items())
alist.sort(key=get_number, reverse=True)
alist[:10]
````
