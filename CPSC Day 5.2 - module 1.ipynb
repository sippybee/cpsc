```python
import pandas as pd
import numpy as np
```

```python
# Create a DataFrame with one column "Numbers" and values from 1 - 10
df = pd.DataFrame(list(range(1,11)), columns=["Numbers"])
```

```python
df["new_col"]=df["Numbers"].apply(lambda x: "Even" if x%2==0 else "Odd")
```

```python
# df.loc[ condition, new_col_name] = expression
df.loc[df["Numbers"]%2==0 , "EvenOdd"] = "Even"
df.loc[df["Numbers"]%2 !=0 , "EvenOdd"] = "Odd"
# one statemnt
```

```python
df["NewEO"] = np.where(df["Numbers"]%2==0 , "Even", "Odd")
```

```python
store1 = pd.DataFrame()
store1["Product1"] = ["Mac","IPad","IPhone","IWatch","TV"]
store1["Price1"] = [2000, 900, 800, 500, 400]
```

```python
store2 = pd.DataFrame()
store2["Product2"] = ["Mac","IPad","IPhone","IWatch","TV"]
store2["Price2"] = [1900, 900, 750, 500, 350]
```

```python
# use np.where and check if Price1 is the Same as Price 
# if yes return True or False save results in 
# store1 df under name "Same"
store1["Same"] = np.where(store1["Price1"]==store2["Price2"], True, False)
```

```python
store1
# if Price1 is greater than Price2 get difference between the prices
# otherwise return 0 ---> "Diff"
store1["Diff"] = np.where(store1["Price1"] > store2["Price2"], store1["Price1"] - store2["Price2"], 0)
# if Col Same False reduce Price1 by 20% if True reduce by 10%
# save under "Sale"
```

```python
store1["Sale"] = np.where(store1["Same"], 0.9*store1["Price1"],0.8*store1["Price1"])
```

```python
store1["NewSale"] = store1.apply(lambda d : 0.9*d["Price1"] if d["Same"] else 0.8*d["Price1"] ,axis=1)
```
