```python
import pandas as pd
```

```python
df = pd.DataFrame()
```

```python
df["Vol"] = ["32.9 K", "56.0 M", "90.0 K", "90.0 M"]
```

```python
df["NV"] = df["Vol"].apply(lambda x : float(x.strip("K"))*1000 if "K" in x else float(x.strip("M"))*1000000)
```

```python
df["NewVol"] = ["32.9 B", "56.0 M", "90.0 K", "90.0 Crazy"]
```

```python
def cleaning(x):
    if "K" in x:
        number = float(x.strip("K"))*1000 
    elif "M" in x:
        number = float(x.strip("M"))*1000000
    elif "B" in x:
        number = float(x.strip("M"))*1000000000000
    else:
        number = 0
    return number
```

```python
df["NewVol"].apply(cleaning)
```
