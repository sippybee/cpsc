````python
check = input("How much is your meal? ")
# print(check, type(check))
check_float = float(check)
# print(check_float, type(check_float))
tax_rate = 0.08
tip_rate  = 0.20

tax = check_float * tax_rate
tip = check_float * tip_rate
total = check_float + tax + tip
print("Your total is ${:.2f}".format(total))
print(f"Your total is ${total:.2f}")
````

````python
# Prices
tshirt_price = 20.00
hat_price  = 25.00

#inputs
tshirt_amnt = input("How many tshirts? ")
hat_amnt  = input("How many hats? ")

tshirt_amnt_int = int(tshirt_amnt)
hat_amnt_int  = int(hat_amnt)

total_tshirts = tshirt_amnt_int * tshirt_price
total_hats = hat_amnt_int * hat_price
grand_total = total_tshirts + total_hats
print(grand_total)
````

````python
shopping_list = ["bananas","apples","bread"]
item = input("What else should I get? ")
shopping_list.append(item)

for food in shopping_list:
    if food == "beer":
        age = input("How old are you?")
        if age.isdigit():
            age_int = int(age)
            if age_int > 21:
                print("Buy a beer")
            else:
                print("No beer today")
                shopping_list.remove("beer")
    else:
        print(f"Do not forget to buy {food}")
````

````python
string = "a1p2p3l4e5$$$$@@@@"
````

````python
letters = []
counter_alpha = 0
counter_digit = 0
for i in string:
    if i.isalpha():
        letters.append(i)
        counter_alpha += 1
    elif i.isdigit():
        counter_digit += 1
print(counter_alpha, counter_digit)
"".join(letters)
````

````python
# ask user a letter
# check if letter is a vowel
# ----> vowel
# else consonant
# print ----> consonant
# vowels = ["a","e","i","o","u"]
vowels = "aeiou"

letter = input("Give me a letter ")
if letter.isalpha() and len(letter)==1:
    letter = letter.lower()
    if letter in vowels:
        print(f"{letter} is a Vowel")
    else:
        print(f"{letter} is NOT a Vowel")
else:
    print("Not a valid input")
````
